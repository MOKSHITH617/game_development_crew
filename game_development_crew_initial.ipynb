````python
%%writefile main.py
### Paste your code below
# python
import pygame
import sys
import random

# --- Game Constants ---
# Screen dimensions
WIDTH = 800
HEIGHT = 400
GROUND_HEIGHT = 50
GROUND_LEVEL = HEIGHT - GROUND_HEIGHT

# Colors
WHITE = (255, 255, 255)
BLACK = (0, 0, 0)
GREEN = (0, 255, 0)  # Player color
RED = (255, 0, 0)    # Obstacle color
BLUE = (135, 206, 235) # Sky background color
BROWN = (139, 69, 19) # Ground color

# Game parameters
FPS = 60
SCROLL_SPEED_INITIAL = 5          # Initial speed of the game world
SCROLL_SPEED_ACCELERATION_RATE = 0.005 # How much speed increases per score unit
MAX_SCROLL_SPEED = 15             # Maximum allowed scroll speed

# Player parameters
PLAYER_WIDTH = 20
PLAYER_HEIGHT = 40
PLAYER_START_X = 50               # Player's fixed horizontal position

JUMP_VELOCITY_INITIAL = -15       # Initial upward velocity when jumping
GRAVITY = 0.8                     # Gravity applied to the player

# Obstacle parameters
OBSTACLE_MIN_WIDTH = 20
OBSTACLE_MAX_WIDTH = 70
OBSTACLE_HEIGHT = 30              # All obstacles have the same height
OBSTACLE_MIN_GAP = 200            # Minimum horizontal distance between obstacles
OBSTACLE_MAX_GAP = 400            # Maximum horizontal distance between obstacles

# --- Pygame Initialization ---
pygame.init()
pygame.font.init() # Initialize font module for text rendering

screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Pixel Runner")
clock = pygame.time.Clock() # To control game speed/FPS

# Fonts for text display
font = pygame.font.Font(None, 36)          # Default font, size 36 for score
game_over_font = pygame.font.Font(None, 72) # Larger font for "GAME OVER" message

# --- Game State Variables ---
scroll_speed = SCROLL_SPEED_INITIAL # Current scrolling speed
score = 0.0                         # Player's current score (using float for smoother accumulation)
game_over = False                   # Flag to indicate if the game is over

# Player state
player_x = PLAYER_START_X
player_y = GROUND_LEVEL - PLAYER_HEIGHT # Player starts on the ground
player_y_velocity = 0               # Player's vertical velocity
on_ground = True                    # True if player is on the ground
# Rect object for player for drawing and collision detection
player_rect = pygame.Rect(player_x, player_y, PLAYER_WIDTH, PLAYER_HEIGHT)

# Obstacle list
obstacles = []

# Ground for continuous scrolling illusion
# We use two ground segments to create a seamless loop
ground_image = pygame.Surface((WIDTH, GROUND_HEIGHT))
ground_image.fill(BROWN) # Fill ground with brown color
ground_x1 = 0    # X-coordinate for the first ground segment
ground_x2 = WIDTH # X-coordinate for the second ground segment, positioned after the first

# --- Helper Functions for Drawing ---

def draw_player(surface, rect):
    """Draws the player character as a green rectangle."""
    pygame.draw.rect(surface, GREEN, rect)

def draw_obstacle(surface, rect):
    """Draws an obstacle as a red rectangle."""
    pygame.draw.rect(surface, RED, rect)

def draw_ground(surface, x1, x2, ground_surface):
    """Draws the two scrolling ground segments."""
    surface.blit(ground_surface, (x1, GROUND_LEVEL))
    surface.blit(ground_surface, (x2, GROUND_LEVEL))

def display_score(surface, score_value):
    """Displays the current score on the top right of the screen."""
    score_text = font.render(f"Score: {int(score_value)}", True, BLACK)
    surface.blit(score_text, (WIDTH - score_text.get_width() - 10, 10))

def display_game_over(surface, final_score):
    """Displays the 'GAME OVER' message and final score in the center of the screen."""
    game_over_text = game_over_font.render("GAME OVER", True, BLACK)
    score_text = font.render(f"Final Score: {int(final_score)}", True, BLACK)

    # Get rectangles for centering the text
    game_over_rect = game_over_text.get_rect(center=(WIDTH // 2, HEIGHT // 2 - 30))
    score_rect = score_text.get_rect(center=(WIDTH // 2, HEIGHT // 2 + 20))

    surface.blit(game_over_text, game_over_rect)
    surface.blit(score_text, score_rect)

# --- Main Game Loop ---
running = True
while running:
    # --- Event Handling ---
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False # Exit the game loop
        if event.type == pygame.KEYDOWN:
            # Check for jump input (Spacebar or Up arrow)
            if (event.key == pygame.K_SPACE or event.key == pygame.K_UP):
                # Player can only jump if on the ground and game is not over
                if on_ground and not game_over:
                    player_y_velocity = JUMP_VELOCITY_INITIAL # Apply upward velocity
                    on_ground = False # Player is now in the air

    # --- Game Update Logic (only if game is not over) ---
    if not game_over:
        # Update player's vertical movement
        player_y_velocity += GRAVITY # Apply gravity
        player_y += player_y_velocity # Update player's Y position

        # Keep player on the ground (prevent falling through)
        if player_y >= GROUND_LEVEL - PLAYER_HEIGHT:
            player_y = GROUND_LEVEL - PLAYER_HEIGHT # Reset to ground level
            player_y_velocity = 0                   # Stop vertical movement
            on_ground = True                        # Player is now on the ground

        # Update player_rect's Y position to match player_y for drawing and collision
        player_rect.y = player_y

        # Increase scroll speed gradually based on score, up to a maximum
        scroll_speed = min(SCROLL_SPEED_INITIAL + (score * SCROLL_SPEED_ACCELERATION_RATE), MAX_SCROLL_SPEED)

        # Update ground positions for scrolling illusion
        ground_x1 -= scroll_speed
        ground_x2 -= scroll_speed

        # If a ground segment moves completely off-screen, reposition it to the right of the other segment
        if ground_x1 < -WIDTH:
            ground_x1 = ground_x2 + WIDTH
        if ground_x2 < -WIDTH:
            ground_x2 = ground_x1 + WIDTH

        # --- Obstacle Generation ---
        # Generate a new obstacle if:
        # 1. There are no obstacles yet (spawn the first one immediately off-screen)
        # 2. Or, the right edge of the last existing obstacle is far enough to the left
        #    to allow for a new obstacle to be spawned with the desired gap.
        if not obstacles or obstacles[-1].right < WIDTH - random.randint(OBSTACLE_MIN_GAP, OBSTACLE_MAX_GAP):
            obstacle_width = random.randint(OBSTACLE_MIN_WIDTH, OBSTACLE_MAX_WIDTH)
            # Create a new obstacle Rect at the right edge of the screen, sitting on the ground
            new_obstacle = pygame.Rect(WIDTH, GROUND_LEVEL - OBSTACLE_HEIGHT, obstacle_width, OBSTACLE_HEIGHT)
            obstacles.append(new_obstacle)

        # Move existing obstacles to the left
        for obstacle in obstacles:
            obstacle.x -= scroll_speed

        # Remove obstacles that have moved completely off-screen to the left
        obstacles = [obstacle for obstacle in obstacles if obstacle.right > 0]

        # Collision Detection: Check if player collides with any obstacle
        for obstacle in obstacles:
            if player_rect.colliderect(obstacle):
                game_over = True # Set game_over flag to True on collision
                break # No need to check other obstacles

        # Update Score: Score increases continuously based on current scroll speed
        score += scroll_speed / FPS

    # --- Drawing ---
    screen.fill(BLUE) # Fill background with sky blue color

    draw_ground(screen, ground_x1, ground_x2, ground_image) # Draw the scrolling ground

    draw_player(screen, player_rect) # Draw the player character

    # Draw all active obstacles
    for obstacle in obstacles:
        draw_obstacle(screen, obstacle)

    display_score(screen, score) # Display the current score

    # If the game is over, display the "GAME OVER" message and final score
    if game_over:
        display_game_over(screen, score)

    pygame.display.flip() # Update the entire screen to show drawings
    clock.tick(FPS)       # Control frame rate to FPS

# --- Game Exit ---
pygame.quit() # Uninitialize pygame modules
sys.exit()    # Exit the program

````
